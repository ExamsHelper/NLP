Programs: 4.a) synonyms, 4.b) antonyms

## 4.a) synonyms

```python
'''4.a) synonyms'''

import nltk
nltk.download('wordnet')
from nltk.corpus import wordnet

words = set()
for syn in wordnet.synsets("good"):
    for lemma in syn.lemmas():
        words.add(lemma.name())

print(words)
```

## 4.b) antonyms

```python
'''4.b) antonyms'''

import nltk
nltk.download('wordnet')
from nltk.corpus import wordnet

words = set()
for syn in wordnet.synsets("good"):
    for lemma in syn.lemmas():
        for ant in lemma.antonyms():
            words.add(ant.name())

print(words)
```
